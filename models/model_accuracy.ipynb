```python
import numpy as np
from mock_model import MockModel
import pandas as pd
from sklearn.model_selection import train_test_split
import random
import sklearn
import json
```

```python
def get_scores(dataframe):
    score_columns = [column for column in dataframe.columns if 'label_' in column]
    return dataframe[score_columns].to_numpy()
```

```python
seed = 42
dataset_name = '../data/imagenet/probas200.csv'
mapping_file = '../data/imagenet/id_to_label200.json'

np.random.seed(seed)
random.seed(seed)
test_size_ratio = 0.8 if 'imagenet' in dataset_name else 0.2

id_to_label = {}

with open(mapping_file, encoding='utf-8') as f:
    string_mapping = json.load(f)
    for i, string_node in enumerate(string_mapping):
        id_to_label[i] = string_node
```

```python

scores = pd.read_csv(dataset_name)
# scores["id"] = scores.index
labels = [col.replace('label_', '') for col in scores.columns if col.startswith('label_')]
score_columns = [f"label_{i}" for i in labels] + ['id',]
model = MockModel(
    scores[score_columns],
    "id",
    # classes=labels,
    softmax=False
)

model = model.fit(
    scores['id'],
    scores['label']
)

X_calib, X_test, y_calib, y_test = train_test_split(
    scores[score_columns],
    scores['label'],
    test_size=0.8,
    stratify=scores['label'],
    random_state=42,
)

y_pred = model.predict(X_calib)

print('accuracy:', sklearn.metrics.accuracy_score(
    y_calib,
    y_pred
))
```
